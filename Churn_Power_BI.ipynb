```python
import pandas as pd
import os
import shutil

split_path = (
    "/content/drive/MyDrive/Project_Churn Prediction/"
    "03_Train_Test_Split"
)

evaluation_path = (
    "/content/drive/MyDrive/Project_Churn Prediction/"
    "05_Model_Evaluation"
)

importance_path = (
    "/content/drive/MyDrive/Project_Churn Prediction/"
    "06_Feature_Importance"
)

powerbi_path = (
    "/content/drive/MyDrive/Project_Churn Prediction/"
    "07_PowerBI"
)

os.makedirs(powerbi_path, exist_ok=True)
```

```python
X_test = pd.read_csv(
    f"{split_path}/X_test.csv"
)

y_test = pd.read_csv(
    f"{split_path}/y_test.csv"
).squeeze("columns")

predictions = pd.read_csv(
    f"{evaluation_path}/Best_Model_Predictions.csv"
)

evaluation = pd.read_csv(
    f"{evaluation_path}/Model_Evaluation_Results.csv"
)

feature_importance = pd.read_csv(
    f"{importance_path}/Top_20_Features.csv"
)
```

```python
dashboard_df = X_test.reset_index(drop=True).copy()

dashboard_df.insert(
    0,
    "Customer_Row_ID",
    range(1, len(dashboard_df) + 1)
)

dashboard_df["Actual_Churn"] = (
    y_test.reset_index(drop=True)
)

dashboard_df["Predicted_Churn"] = (
    predictions["Predicted_Churn"].reset_index(drop=True)
)

dashboard_df["Churn_Probability"] = (
    predictions["Churn_Probability"].reset_index(drop=True)
)

dashboard_df["Risk_Category"] = (
    predictions["Risk_Category"].reset_index(drop=True)
)

display(dashboard_df.head())
print("Dashboard rows:", len(dashboard_df))
```

```python
required_columns = [
    "Actual_Churn",
    "Predicted_Churn",
    "Churn_Probability",
    "Risk_Category"
]

print(dashboard_df[required_columns].isnull().sum())

if dashboard_df[required_columns].isnull().sum().sum() > 0:
    raise ValueError(
        "Dashboard dataset contains missing prediction values."
    )

print("Dashboard dataset validated successfully.")
```

```python
dashboard_df.to_csv(
    f"{powerbi_path}/Dashboard_Data.csv",
    index=False
)

evaluation.to_csv(
    f"{powerbi_path}/Model_Performance.csv",
    index=False
)

feature_importance.to_csv(
    f"{powerbi_path}/Feature_Importance.csv",
    index=False
)

print("Power BI files saved successfully.")
```
