```python
import pandas as pd
from sklearn.model_selection import train_test_split
import os
```

```python
df = pd.read_csv(
    "/content/drive/MyDrive/Project_Churn Prediction/02_Feature_Engineering/Telco_Customer_Churn_Feature_Engineered.csv"
)

print("Dataset Loaded Successfully")
print("Dataset Shape:", df.shape)

display(df.head())
```

```python
print("="*60)
print("DATASET INFORMATION")
print("="*60)

df.info()
```

```python
print("Missing Values")

display(df.isnull().sum())
```

```python
X = df.drop("Churn", axis=1)

y = df["Churn"]

print("Feature Matrix Shape :", X.shape)
print("Target Shape :", y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
print("Training Features :", X_train.shape)
print("Testing Features  :", X_test.shape)

print()

print("Training Labels :", y_train.shape)
print("Testing Labels  :", y_test.shape)
```

```python
print("Training Class Distribution")

display(
    y_train.value_counts().to_frame("Count")
)

display(
    round(
        y_train.value_counts(normalize=True)*100,
        2
    ).to_frame("Percentage")
)
```

```python
print("Testing Class Distribution")

display(
    y_test.value_counts().to_frame("Count")
)

display(
    round(
        y_test.value_counts(normalize=True)*100,
        2
    ).to_frame("Percentage")
)
```

```python
output_path = "/content/drive/MyDrive/Project_Churn Prediction/03_Train_Test_Split"

os.makedirs(output_path, exist_ok=True)

print("Output folder created successfully.")
```

```python
train_df = X_train.copy()

train_df["Churn"] = y_train.values

train_df.to_csv(
    f"{output_path}/Training_Dataset.csv",
    index=False
)

print("Training dataset saved.")
```

```python
test_df = X_test.copy()

test_df["Churn"] = y_test.values

test_df.to_csv(
    f"{output_path}/Testing_Dataset.csv",
    index=False
)

print("Testing dataset saved.")
```

```python
X_train.to_csv(
    f"{output_path}/X_train.csv",
    index=False
)

X_test.to_csv(
    f"{output_path}/X_test.csv",
    index=False
)

y_train.to_csv(
    f"{output_path}/y_train.csv",
    index=False
)

y_test.to_csv(
    f"{output_path}/y_test.csv",
    index=False
)

print("Individual train-test files saved.")
```

```python
summary = pd.DataFrame({

    "Dataset":[
        "Training Features",
        "Testing Features",
        "Training Labels",
        "Testing Labels"
    ],

    "Rows":[
        X_train.shape[0],
        X_test.shape[0],
        y_train.shape[0],
        y_test.shape[0]
    ],

    "Columns":[
        X_train.shape[1],
        X_test.shape[1],
        1,
        1
    ]

})

display(summary)
```

```python
summary.to_csv(
    f"{output_path}/Train_Test_Summary.csv",
    index=False
)

print("Summary saved successfully.")
```

```python
import shutil

shutil.make_archive(
    "/content/Train_Test_Split",
    "zip",
    output_path
)

print("ZIP file created successfully.")
```
